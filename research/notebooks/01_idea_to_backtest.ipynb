# 01) Idea → Backtest Request

This notebook helps you turn a research idea into a backend **backtest request** payload.

Outputs are **not production evidence** until you rerun the backtest through the verified pipeline and confirm results.

## 0. Configure API base

Set `API_BASE` to your FastAPI backend.


```python
API_BASE = "http://localhost:8000"  # change if needed
```

## 1. Define the idea

Describe the hypothesis and the strategy constraints you want to test.

```python
idea = {
    "hypothesis": "Example: momentum entries with RSI oversold filter",
    "notes": "Replace with your idea details.",
}
idea
```

## 2. Pick a strategy

This platform runs **registered strategies** via `POST /strategies/backtests`.

We list available strategies and pick one.

```python
import requests

strategies = requests.get(f"{API_BASE}/strategies/", timeout=30).json()
print("Strategies:", len(strategies))
strategies[:3]
```

```python
# Choose a strategy id from the list above
strategy_id = strategies[0]["id"] if strategies else None
strategy_id
```

## 3. Build the backtest config

This config is sent as the `config` object in the backtest request.

```python
backtest_config = {
    "start_date": "2020-01-01",
    "end_date": "2023-01-01",
    "initial_capital": 1_000_000.0,
    "commission_rate": 0.005,
    "slippage_bps": 5.0,
    "execution_delay_bars": 1,
}
backtest_config
```

## 4. Final request payload

This is the payload you will submit in notebook 02.

```python
payload = {"strategy_id": int(strategy_id), "config": backtest_config}
payload
```
